```python
import cv2
from ultralytics import YOLO
```

```python
model=YOLO('yolov8n.pt')
```

```python
%pip install opencv-python
```

```python
# کلاس‌هایی که می‌خواهیم تشخیص بدهیم
selected_classes = [0, 2, 4, 16]

# نام کلاس‌ها
class_names = {
    0: 'Person',
    2: 'Car',
    4: 'Airplane',
    16: 'Dog'
}

# اتصال به دوربین
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Camera is not available!")

else:
    while True:

        ret, frame = cap.read()

        if not ret:
            print("Cannot read camera!")
            break

        # نمایش تصویر مثل آینه
        frame = cv2.flip(frame, 1)

        # تشخیص اشیا
        result = model(
            frame,
            conf=0.25,
            classes=selected_classes,
            verbose=False
        )[0]

        # رسم کادرها
        output = result.plot()

        # شمارش اشیا
        counts = {
            'Airplane': 0,
            'Person': 0,
            'Car': 0,
            'Dog': 0
        }

        # بررسی نتایج
        if result.boxes is not None:

            for box in result.boxes:

                class_id = int(box.cls[0])
                confidence = float(box.conf[0])

                name = class_names[class_id]

                counts[name] += 1

        # نمایش عنوان پروژه
        cv2.putText(
            output,
            "AI DEFENSE MONITORING",
            (10, 30),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 255),
            2
        )

        # نمایش تعداد اشیا
        y = 65

        for name, count in counts.items():

            text = f"{name}: {count}"
            cv2.putText(
                output,
                text,
                (10, y),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

            y += 30

        # نمایش تصویر
        cv2.imshow(
            "AI Defense Monitoring System",
            output
        )

        # خروج با ESC
        if cv2.waitKey(1) & 0xFF == 27:
            break

    cap.release()
    cv2.destroyAllWindows()
```
